```python
wish = 'Hello'
name = input('Enter your name: ')
wish
```

```python
print(wish)
```

```python
message = wish + ' ' + name
print(message)
```

```python
## now concatenate string 3
```

```python
greeting = ', Welcome to Python Programming.'
greeting
```

```python
greeting_message = message + greeting
print(greeting_message)
```

```python
## 2. String slicing and Indexing
```

```python
greeting_message[0] # first character of the string
```

```python
greeting_message[-1] # last character of the string
```

```python
greeting_message[0:5] # first 5 character of the string
```

```python
greeting_message[33:] # last 12 characters of the string (including '.')
```

```python
greeting_message[34:] # last 11 character of the string
```

```python
greeting_message[::-1] # the string in reverse
```

```python
greeting_message[26:32] # print the word python
```
